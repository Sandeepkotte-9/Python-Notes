Variable names must start with a letter (A-Z or a-z) or an underscore (_).
Variable names cannot start with a number.
Variable names can contain letters, numbers, and underscores.
Variable names are case-sensitive (age, Age, and AGE are different variables).
Variable names cannot be Python keywords such as if, for, class, or while.

```python
Type casting in Python is the process of converting a value from one data type to another. It is useful when you need to perform operations that require a specific data type.

There are two types of type casting:

Implicit Type Casting (Automatic conversion by Python)
Explicit Type Casting (Manual conversion by the programmer)
```

```python
#Implicit Type Casting
#Python automatically converts a smaller data type to a larger compatible data type to avoid data loss.
num1 = 10      # int
num2 = 5.5     # float

result = num1 + num2

print(result)
print(type(result))
```

```python
#Explicit Type Casting
#In explicit type casting, you manually convert a value using built-in functions such as int(), float(), str(), or bool().
num = "100"

number = int(num)

print(number)
print(type(number))
```
