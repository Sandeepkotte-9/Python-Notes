Python Error Handling:
When an error occurs, or exception as we call it, Python will normally stop and generate an error message. There are some blocks to handle these exceptions, as follows:

The try block lets you test a block of code for errors.

The except block lets you handle the error.

The else block lets you execute code when there is no error.

The finally block lets you execute code, regardless of the result of the try- and except blocks.

```python
try:
    #f = open("test.txt")        # Will give us error
    f = open(r"/Users/sandeepkotte/Documents/Python Notes/test.txt", "w")
    try:
        f.write("Everything good so far")
    except:
        print("There is an error writing into the file")
    else:
        print("Yeah! No error")

    finally:
        f.close()
except:
    print("File not found")
    
# import os
# if os.path.exists("test.txt"):
#   os.remove("test.txt")
# else:
#   print("The file does not exist")
```

File Handling:
File handling is an important part of any web application. Python has several functions for creating, reading, updating, and deleting files.

Whenever we need to work with files, we have to open the file and so in Python we use the open() function. The open() function takes two parameters; filename, and mode. There are four different methods (modes) for opening a file:

"r" - Read - Default value. Opens a file for reading, error if the file does not exist

"w" - Write - Opens a file for writing, creates the file if it does not exist One of the disadvantages of write mode in file is that, whenever we'll try to open the file with write mode, it'll just overwrite and remove all the previous contents. Just to overcome this, one more mode was introduced, which is append.

"a" - Append - Opens a file for appending, creates the file if it does not exist

```python
# Working with the text files.

# Writing to a text file:

with open(r"/Users/sandeepkotte/Documents/Python Notes/test.txt", "w") as t:
    t.write("This file is only to under File handling in Python.")
    t.write("\nLet's try to deep dive into the File handling concepts")
    t.write("\nIt can be a little boring however once you understand, it'll be fun! LOL!")
    t.write("\nJust keep in mind, we have open(), write, read and append modes in file")
    t.write("\nThat's all, see you next time!")    
    
# Reading from a text file:

with open(r"/Users/sandeepkotte/Documents/Python Notes/test.txt", "r") as t:
    summary = t.read()
    print("Here are the demo file Contents: ")
    print(summary)
```

```python
# Append mode in file handling:

# Let's try to open again the previous file with write mode and then read it:

with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "w") as f:
    f.write("Hey, you just deleted the previous contents of this file!")
    
with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "r") as f:
     print(f.read())
    
# As we mentioned, the only disadvantage of write mode is till overwrite the previous data, thus let's see the append mode, using which we can keep the original contents of the file and add new contents to it.

with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "a") as f:
    f.write("\nTake care!")

with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "r") as f:
    print(f.read())  
```

```python
# strip() removes the white spaces and new lines from both ends of the lines. In the original line, we can see lot of white spaces at the start of line 3 and 4, however using strip() those are removed and then the output is displayed. Note that the white spaces after last word and the full stop is not removed, because strip() does not removes the spaces within the words.

with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "r") as t:
    for line in t:
        print("Line: ", line.strip())
```

```python
# Copying the file contents: To do this, we'll open source file in read mode and destination file in write mode. Using for loop we'll try to iterate and write all the contents in the destination file. 

with open(r"/Users/sandeepkotte/Documents/Python Notes/test.txt", "r") as src, open(r"/Users/sandeepkotte/Documents/Python Notes/copy_test.txt", "w") as dest:
    for line in src:
        dest.write(line)
        
with open(r"/Users/sandeepkotte/Documents/Python Notes/copy_test.txt", "r") as f:
    data = f.read()
    print("Copy contents: ")
    print(data)
```

```python
# seek(): we can give a line number as a parameter and then write some other statement on that line number. r+ will open the file in read and write mode simultaneously.

with open(r"/Users/sandeepkotte/Documents/Python Notes/demofile1.txt", "r+") as f:
    message = f.read()
    f.seek(300)
    f.write("This is the end!")
    
    print(message)
```

However if we open the file using with statement, then there is no need to close the file, as with statement takes care of it. with statement is also known as context managers in Python, as those statements are the one, which comes with an entry and exit statements.

Modules:
Python has a way to put definitions in a file and use them in a script or in an interactive instance of the interpreter. Such a file is called a module; definitions from a module can be imported into other modules or into the main module. A module is a file containing Python definitions and statements. The file name is the module name with the suffix .py appended. In short, a single .py file is a module.

Libraries:
A collection of multiple .py files, that is collection of reusable modules, functions and classes is a library. It allows the developers to perform complex tasks without writing the code from the scratch.

Packages:
Packages are the collection of modules and libraries. It contains several .py file, usually signaled by an init.py file and mostly in Python we'll use command as pip install to install such packages. Pandas is one of the Python libraries and packages too.
