Range():

The built-in range() function returns an immutable sequence of numbers, commonly used for looping a specific number of times. This set of numbers has its own data type called range. Let's some of the examples of range().

```python
#Print numbers from 1 to 5

for i in range(1, 6):
    print(i)
```

```python
# If we wanted to create a list of numbers:

print(list(range(9)))

# If we wanted in the range function we can add the starting parameter, ending parameter and a step size too:

print(list(range(3, 9)))  

print(list(range(3, 15, 5)))   #start with 3 and then increement with 5 and go at the max till 15
```

```python
names = ["Sandy", "Ram", "Ray"]
for n in names:
    print(n)
    
```

```python
#Iterate through a list

fruits = ["Apple", "Banana", "Orange"]

for fruit in fruits:
    print(fruit)
```

#while Loop
Definition

A while loop repeatedly executes a block of code as long as a specified condition is True. The loop stops when the condition becomes False.

while condition:
    

```python
#Print numbers from 1 to 5

count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
#Countdown

count = 5

while count > 0:
    print(count)
    count -= 1

print("Done!")
```

The Break statement: If we use a break statement in a loop then it'll come out of the nearest loop.

The continue statement: The loop will continue however it'll skip the execution of the loop when the condition matches. Below are the examples on it:

```python
for i in range(10):
    if i%3 == 0:
        continue 
    print(i)
```

Digit pattern analyser: You are given an integer n. Your task is to analyze all numbers from 1 to n and:
Print numbers that have exactly 2 digits
For those numbers, also print the sum of their digits
Skip numbers where the digits are the same (like 11, 22, 33...)
Example idea: If n = 50, you would process numbers like 10–50 but skip 11, 22, 33, 44.

n1 = int(input("Enter upper limit (2-digit range): "))

for i in range(10, min(n1 + 1, 100)):
    r = i % 10
    d = i // 10

    if r == d:
        continue

    print(f"Number: {i}, Sum of digits: {r + d}")

```python
Running Balance Simulator (Condition + Stop Rule): You start with a balance of 100. You are given a list of transactions (positive = deposit, negative = withdrawal).
Rules:

Apply each transaction one by one
If balance ever drops below 50, stop immediately
Print final balance and how many transactions were processed
```

```python
bal = 100
transactions = [20, -30, -40, 10, -5]

count = 0

for t in transactions:
    bal += t
    count += 1

    if bal < 50:
        break

print(f"Final Balance: {bal}")
print(f"Transactions processed: {count}")
```

Collatz-Like Sequence Tracker (Classic but non-trivial loop use):
Start with a number n. Apply the rules:

If n is even → n = n / 2

If n is odd → n = 3n + 1

Continue until n becomes 1.

Your task:

Print each step
Count how many steps it takes to reach 1
Example: If n = 6 → sequence continues until it reaches 1.

```python
numm = int(input("Enter a number: "))

step = 0

while numm != 1:
    print(numm)

    if numm % 2 == 0:
        numm = numm // 2
    else:
        numm = 3 * numm + 1

    step += 1

print(1)
print("Steps to reach 1:", step)
```
