```python
# Functions in Python:  A function is a block of code which only runs when it is called. A function can return data as a result. A function helps avoiding code repetition.

def add(n1, n2):            # function defination, n1 and n2 are the function input parameters
    addition = n1 + n2      # function body
    return addition         # function return statement

res = add(5, 6)              # calling of function
print(res)
```

```python
# Recursive function: A function calling function

def fact(n):
    if n == 0:
        return 1
    
    else:
        factorial = n * fact(n-1)       # the fact(n) is calling fact(n-1), thus recursive function
    return factorial

num = int(input("Please enter a number whose factorial you wanted to know: "))
result = fact(num)
print(f"The factorial of {num} is: {result}")
```

#Generators:
With the iterators, we can iterate over the already available data, however if we don't have a given data or we need to generate some data. In that case we need to use generators.

Generators are special type of iterators that is written like a functions but uses yield keyword instead of return. It pauses when it yields and resumes from where it let off. It is used to generate values on fly, saving memory.

Why yield instead of return?
return() ends the function and sends back a value, yield pauses the functions and remembers its state, allowing it to resume later

With the yeild keyword it will store the previous values when we try to ask for the new values. For ex. if we are trying to generate cubes of numbers from 1 to 10 and if we wanted to get the output for number 3, then the yeild functions will try to store or remember the previous generated values which are 1, 2, and for the 3 it'll generate the value and remember it. Later in future if we try to access for number 5, then it won't re-generate it, it'll just bring it back and show us and only generate for 4 and 5.

```python
# Creating a generators:

def gen(n):
    for x in range(n):
        yield x
    
gen(9)      # This is to create a generator object, on the console, we will see this: <generator object gen at 0x107824110>

# for i in gen(10):     # this is to print the values
#     print(i)
```

Python Iterators:
An iterator is an object that contains a countable number of values. An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods iter() and next().

Why to use iterators:
Lazy evaluation: When we use iterators, we use next() to get the output or to iterate, thus it is kind of on demand processing.

Memory efficiency: Whenever we create a list, tuple and iterate it using a for loop and memory is allocated at the start. When the data is small, it is fine, however when we have big data, in that case, using for loop to iterate will consume a lot of memory, so in that case, we can try using iterators and generate the output, as per our requirements. Basically iterators will maintain the current state and it's reference to the next item, making it more memory efficient.

```python
# Creating an iterator:

def display():
    for i in range(1, 5):
        print(f"Hello {i}")
        yield i
        
res = display()
print(res)

print(next(res))
print(next(res))
print(next(res))
print(next(res))
#print(next(res))        # We'll get an error as Stop iteration, because we have specified only a range of 1 to 5.
```

```python
# Iterator vs Iterable: Lists, tuples, dictionaries, and sets are all iterable objects. They are iterable containers which you can get an iterator from. All these objects have a iter() method which is used to get an iterator:

# 1. With a list:
list1 = ["Python", 1, 4, "Data Science"]
iter1 = iter(list1)     # using iter keyword, we can change the list which is iterable to iterators

print(next(iter1))      # output will be Python
print(next(iter1))      # output will be 1
print(next(iter1))      # output will be 4
print(next(iter1))      # output will be Data Science

# 2. With a string:

name1 = "Sephora"
iter2 = iter(name1)

print(next(iter2))      # output will be S
print(next(iter2))      # output will be e
print(next(iter2))      # output will be p
print(next(iter2))      # output will be h
print(next(iter2))      # output will be o
print(next(iter2))      # output will be r
print(next(iter2))      # output will be a
#print(next(iter2))      # Error, as the stop iteration
```

```python
# 3. With a tuple:

tup1 = ("Cloudy", "Rain", "Sun")
iter3 = iter(tup1)

print(next(iter3))      # output will be Ray
print(next(iter3))      # output will be Rain
print(next(iter3))      # output will be Sun
#print(next(iter3))      # Error, as the stop iteration
```

map():
It is like mapping function to a list of items. So basically one to multiple relation. Mostly we will use map() when we wanted to create a temporary function, take it's output and later that functions will neverbe used.

It is in the form of map(function, sequence)
The map() function executes a specified function for each item in an iterable. The item is sent to the function as a parameter.

function: The function to apply to each item.
iterable: The sequence (list, tuple, etc.) whose items will be processed.

```python
#Square each number

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * x, numbers)

print(list(result))
```

```python
#Convert strings to uppercase
names = ["alice", "bob", "charlie"]

result = map(str.upper, names)

print(list(result))
```

```python
#Convert strings to integers

numbers = ["10", "20", "30", "40"]

result = map(int, numbers)

print(list(result))
```

```python
#Add 10 to each number

numbers = [5, 10, 15, 20]

result = map(lambda x: x + 10, numbers)

print(list(result))
```

```python
#Find the length of each word

words = ["Python", "Java", "C++", "JavaScript"]

result = map(len, words)

print(list(result))
```
