Object:In Python, everything is an object. An object is an instance of a class that contains: Data (Attributes) – variables that store information. Methods (Functions) – actions that the object can perform.

```python
##Creating an an Class and object
##What is self?

##self refers to the current object (instance) being created or used.
##It allows the object to access its own variables and methods.

##__init__ is a constructor method in Python.It is automatically called when an object is created.The purpose of __init__ is to initialize the object's attributes.

##Store the value of name inside the current object's name attribute.Without self, Python wouldn't know which object's data you're referring to.


class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def display(self):
        print("Name:", self.name)
        print("Role:", self.role)

emp1 = Employee("Rishi", "Java Developer")

emp1.display()

##Employee = Class
##emp1 = Object
##name and role = Attributes
##display() = Method
```

```python
##Methods:A method is a function that belongs to an object or a class.
##It defines the behavior of that object and is called using the dot (.) operator.


##Types of Methods in Python
##1. Instance Methods

##These work with object data and use self.


class Student:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(self.name)

s1 = Student("Sandeep")
s1.show()
```

```python
##2. Class Methods

class Student:
    school = "ABC School"

    @classmethod
    def get_school(cls):
        return cls.school

print(Student.get_school())
```

```python
##3. Static Methods

class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

What is str() in Python?
str() is a special (magic/dunder) method used to define how an object should be represented as a string when you use: .print(object) .str(object)

Without str(), Python displays a default object reference that is not very user-friendly.The str() method is used to define how an object should be displayed when you print it.

```python
##Example Without __str__()

class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("John")

print(s1)

print(s1)   ##This output is not very meaningful.
```

```python
##Example With __str__()
##Now the object is displayed in a readable format.

class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return "Student Name: " + self.name

s1 = Student("John")

print(s1)
```

##__repr__() is another special method in Python.

##What is __repr__()?
##__repr__() is a special method that defines the official string representation of an object. It is called by repr() and is mainly used for debugging and development purposes.

```python
## with __ repr__

class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return "Student('" + self.name + "')"

s1 = Student("John")

print(repr(s1))

##How it works
##repr(p1) calls the __repr__() method.
##__repr__() returns the string "Person('John')".
```

```python
## Without __repr__()
class Person:
    def __init__(self, name):
        self.name = name

p1 = Person("John")

print(repr(p1))     ##Since __repr__() is not defined, Python shows the default object representation (memory address).
```

Abstraction:Hiding unnecessary details
In simple words, the user knows what an object does, but not how it does it. Abstraction is the process of hiding implementation details and showing only essential features to the user.

Real-Life Example
You know how to start and drive the car. You don't need to know how the engine works internally.

This is abstraction.

```python
#### Simple Example

from abc import ABC, abstractmethod

class Remote(ABC):

    @abstractmethod
    def turn_on(self):
        pass

class TV(Remote):

    def turn_on(self):
        print("TV is ON")

tv = TV()
tv.turn_on()
```

Inheritance:
Inheritance is an OOP (Object-Oriented Programming) concept where one class acquires the properties and methods of another class.

The existing class is called the Parent Class (or Base Class).

The new class is called the Child Class (or Derived Class,sub class).

Why Use Inheritance?
Reuse code.

Avoid writing the same code multiple times.

Make programs easier to maintain.

```python
## Simple Example

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass

d = Dog()
d.sound()
```

What are *args and **kwargs in Python?
*args is used to pass multiple positional arguments to a function as a tuple, while **kwargs is used to pass multiple keyword arguments as a dictionary.

They are used to pass multiple arguments to a function.

*args (Non-keyword arguments) Used to pass any number of values Stored as a tuple

**kwargs (Keyword arguments) Used to pass key-value pairs Stored as a dictionary

```python
##*args (Non-keyword arguments)
## example 

def add(*args):
    total = 0
    for num in args:
        total += num
    print(total)

add(10, 20, 30)
add(5, 15)

##Explanation
##You can pass any number of values.
##*args collects them like a tuple:
(10, 20, 30)
(5, 15)
```

```python
##**kwargs (Keyword arguments)
## example 

def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(name="John", age=20, city="Boston")

##Explanation
##You pass data in key=value format
## Python stores it as a dictionary:
```

##*args → Many values → like a list/tuple
##**kwargs → Many key-value pairs → like a dictionary

Method overriding:
Method overriding happens when a child class provides its own implementation of a method that is already defined in the parent class.

In simple words:Same method name in parent and child class, but different behavior in child class.

```python
## Example:

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

a = Animal()
a.sound()

d = Dog()
d.sound()
```

Encapsulation :Is one of the Object-Oriented Programming (OOP) concepts.
It means:Binding data (variables) and methods (functions) together into a single unit (class) and restricting direct access to some data.

Encapsulation helps:

Protect data from accidental modification. Improve security. Control how data is accessed and updated.

```python
## Example

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

s1 = Student("Sandeep", 90)

s1.display()
```

What is Dynamic Extension in Python?
Dynamic Extension is a feature of Python that allows attributes and methods to be added to objects or classes during runtime. This provides flexibility and makes Python a highly dynamic language.

```python
## Example1

class Student:
    pass

s1 = Student()

s1.name = "Sandeep"

print(s1.name)
```

```python
##Example 2

class Employee:
    pass

emp = Employee()

emp.name = "Sandeep"
emp.role = "Developer"

print(emp.name)
print(emp.role)

##Explanation:The Employee class has no attributes defined, but we dynamically added name and role while the program was running.
```

```python
##Example3

class Student:
    pass

def show():
    print("I am a Student")

s1 = Student()

s1.show = show

s1.show()

##Explanation:We created a function show() and attached it to the object s1 at runtime.
```

Polymorphism in Python
Definition:Polymorphism means "many forms." The same method can behave differently depending on the object calling it.

```python
## example:

class Dog:
    def sound(self):
        print("Bow Bow")

class Cat:
    def sound(self):
        print("Meow Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()

## Both classes have a method called sound().

## The same method name behaves differently for different objects.

## This is called Polymorphism.
```
