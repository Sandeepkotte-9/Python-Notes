Lists: it is hetrogenous and it is multable Why use list? indexing and slicing with lists mutability and immutability append(), pop()-LIFO, reverse(), sort() Nesting lists, basically matrix List comprehension extend(), insert()

Tuples: exactly same like list, however tuple is immutable We'll use tuple where we don't want the data to be changed ever then we will use tuple

Sets: ordered collection of unique elements, it doesn't allow duplicates discard(), remove(), clear(), pop() - FIFO

Dictionary: it is a key-value pair storage or mappings making a dictionary ways to create a dictionary, using dict keyword Nesting with dictionary dictionary methods dictionary comprehensions

```python
# Creating a dictionary (key-value mapping):

stud = {1 : "Sandy",
        2 : "Mike",
        3 : "Sam"}

# Displaying the dictionary and it's items

print(stud)
print(stud[1])

# Checking length and type:

print(type(stud))
print(len(stud))

print(stud.keys())      # displaying the keys in the given dictionary
print(stud.values()) 
```

```python
# Dictionary Comprehensions:

powers = {x: x ** 4 for x in range(1, 5)}
print(powers)
```

```python
# updating the items in dictionary:

# 1. directly using the key and assigning the value to it:

# stud[3] = "Mike"
#print(stud)

# 2. update():

stud.update({2: "God"})
print(stud)
```

```python
# Nested dictionary:

dict1 = {"film1": "Odyssey",
         "film2": "300",
         "film3" : "Anaconda"}

dict2 = {"year1" : 1997,
         "year2" : 2006,
         "year3" : 1995}

dict3 = {"film_name": dict1, "r_year": dict2}

print(dict3)
```

```python
# Creating a list:

# 1. Using the [] brackets:
list1 = [17, 3, "Python", "Machine Learning"]

# 2. Using list comprehensions:

cubes = [x ** 3 for x in range(5)]
even = [n for n in range(10) if n % 2 == 0 ]

# 3. using list() keyword:

num = list((1,2,3))
char = list(("mno"))
```

```python
# Displyaing the list:

print(list1)
print(cubes)
print(type(even))   # checking the type
print(len(list1))   # to check the length of the list
print(char)
```

```python
# nested lists: It is basically a list inside another list

id = [1, 2, 3]
name = ["Alice", "Ben", "Cizze"]
dept = ["CS", "Maths", "History"]

new_list = [id, name, dept]     #this is one of the example of nested list and what we get is a 3X3 dimensional matrix
print(new_list)
```

```python
# List comprehension: It is one of the concise way to create a list. We can create a new list based on the list already give.

# Create a list which displays the number completely divided by 3 within range 1 to 20

# list_3 = []
# for i in range(1,20):
#     if i % 3 == 0:
#         list_3.append(i)
# print(list_3)

# The above code can be replaced using List comprehension by just writing a single line of code. In the below example I'm replacing the above logic and creating a completely new list using List comprehension. It saves the line of code.

list2 = [m for m in range(1, 20) if m % 3 == 0] 
print(list2)

# Let's take the already created list and using list comprehension, try to make a list out of it.

new_list1 = [c[0] for c in new_list]    # We are trying to access the zeroth column of the matrix new_list
print(new_list1)
```

```python
# Accessing the list items:

print(list1[0])
print(list1[-2])
print(name[1:2])

# Checking if particular item exists in list

if "English" in dept:
    print("Yes")
else:
    print("No")
```

```python
# Changing the items in list:

list1[2] = "SQL"    # Python will be replaced by "SQL"
print(list1)

# insert() to insert a new item in the list

name.insert(3, "David")
print(name)

# Adding the items using append()

dept.append("Science")
print(dept)

# Changing the items in a range:
id[2:4] = [5, 6]
```

```python
# pop(): to remove an item from the list

# name.pop()      # List in first out, so the last added item is removed
# print(name)

# remove(), will remove the item, which is passed a parameter

dept.remove("CS")
print(dept)
```

```python
# Sorting of the list:

dept.sort()
print(dept)

name.sort(reverse = True)       # To sort in descending order
print(name)
```

```python
# Creating a set

# 1. using cruly brackets

set1 = {"White", "Black", "Brown", "Brown", "Blue"}  
set2 = {2, 4, 6, 8, 10, 10, 10, 8}  
set3 = {"Lily", "Siya", 2, 4, 3, 3, "Lily", "Jon", 7, 7, "Jon"}

# 2. using set()

cars = ["BMW", "Audi", "Audi", "Toyota"]
cars_set = set(cars)
```

```python
# Display the set and we can see the duplicates will be removed

print(set1)

print(set2)
print(set3)

print(type(set1))
print(type(set2))

print(type(cars))
print(f"cars: {cars}")

print(type(cars_set))
print(f"cars_set: {cars_set}")

print(len(cars_set))
```

```python
# Accessing the set items:

# 1. Using for loop

for s in set3:
    print(s)
```

```python
# Checking is an item is available in the set or not:

print(6 in set2)
print("Audi" in cars_set)
print("Chewy" in cars_set)
```

```python
# adding the items to the set:

# cars_set.add("Honda")
print(cars_set)
```

```python
# Removing the items from set

set3.pop()      # the first item is removed from here that is FIFO
print(set3)

# set1.remove("Black")
# set2.discard(10)

print(f"set1 {set1}")
print(f"set2 {set2}")
```

```python
# Creating a tuple:

# 1. Using () brackets
tup1 = (8, 9, 3, 5)
tup_n = ("Ploy", "Ray", "Sun", "Tay")
tup_m = (1, 1, "New", 5, 8, 7, "Magic")

# 2. In case we have a list and we wanted to make it a tuple then Using tuple() keyword we can achieve that

listings = ["Peace", 10, 78, "Tango"]
tup2 = tuple(listings)
```

```python
# Displaying the tuple:

print(tup1)

print(tup_n)
print(type(tup_n))      #checking the datatype 
print(len(tup_n))      #checking the length 

print(tup_m)
print(type(tup_m))
print(len(tup_m))

print(f"Given list {listings}")
print(type(listings))

print(f"Changed to tuple {tup2}")
print(type(tup2))
```

```python
# accessing the items in tuple:

print(tup1[3])
print(tup_n[-2])
print(tup_m[2:5])

# Checking if the item exists in the tuple

if "abc" in tup_n:
    print("Got you")
else:
    print("Where are you?")
```
