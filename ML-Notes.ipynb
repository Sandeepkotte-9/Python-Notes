What is ML and why do we need ML?
Machine learning in simple language is training a machine or model to think like a human.
Why to do that? Consider, if we have a huge volume of data and if we need to identify the pattern or understanding of the data, so in that case a human can’t do it, so we’ll need to train a model which can do this on behalf of human, that’s why ML evolved here.

Ways to train the model:
Supervised models: A human will label the data and then using that we’ll start learning the pattern and later when a new information comes, the model will predict it.

Unsupervised models: You directly give the data to the model and it’ll try to classify the data and we will never label the data.

Semi Supervised models: From the 100% data, we’ll label for some % of data and for the rest % of the data, the model will figure out the pattern.

Reinforcement learning: It absorbs the nature and then analyze the data, once it predict it and it is correct, it is rewarded and if the prediction is wrong, then it is penalized.



Types of Supervised models:
Regression: Trying to predict continuous values. Ex. What is the height of the person, weight of person, price predictions

Classification: It’ll classify the data. Spam or not spam, cancer or not cancer

It has :
Binary classification
Multi label classification
Multi class classification


ML workflow:
Start -> Data selection -> Data description (telling a story of the data) -> performing statistical and graphical data analysis -> Data transformation -> Selection on ML model based on EDA -> data standardization and normalization -> creation of train and test data sets -> model training using the ML model -> cal model accuracy -> Hyper parameter training -> saving the created model file -> deployment strategies (live, stream or batch) -> production deployment and testing -> finalizing the retraining approach -> logging and monitoring -> dashboard for monitoring and logging reports ->
Stop

Linear Regression:
Quick notes on this: - For this we have a residual error, it means the difference between actual value and predicted value. - To do that, we have 2 methods:
OLS (Ordinary Least Square) technique: Minimum of residual error. We don’t have a control on that value
Gradient descent: Optimizing technique, which is used to reduce the lost function. To go to the global minima, we apply learning rate. We should start with smaller steps.
Types of cost functions:
Mean Absolute Error (MAE): prone to outliers
Root Mean Squared error (RMSE)
R- squared: provides a measure of fit. RSS (Residual sum of square), TSS (Total Sum of square) and ESS (Explained Sum of Square)
R squared = tss-rss / tss = 1- (rss/tss)

Linear regression is a method used to predict one value based on another by finding the best-fit straight line through the data points.

The model will be as follows:
Trained
Overfit: Model will learn the noise also and it’ll overly perform on train data and will be terribly perform on test data. To avoid this, we need to clean the data properly.
Underfit: The model will be too simple and data is too complex. It’ll not able to learn all the patterns in the data.
Class Imbalance example if we have a data with 80% of spam data and 20% of not spam data, so in this the prediction will not be perfect.

Multi colinarity:

Two independent variables are co-related. Example here is years of experience will give us information on person’s salary and so we’ll not take age variable into consideration. Why we should care about this because in linear regression y = mx + c, so using this equation, if x increases, y will also increase. Thus when we have some data where there are x1, x2, then increase in both of them will impact your values too. So we need to think about this and try to reduce it.
Note: If the learning rate is slow and it is stuck in local minima then in that case, we’ll add some of the momentum

Bias and Variance in ML:
Bias refers to the error caused by oversimplifying a model. High bias means the model oversimplifies, misses the pattern and underfits the data however Low bias is that the model captures patterns well and is closer to the true values.
Variance refers to the error from making the model too sensitive to training data and it captures noises in data too. It fails to give prediction on unseen new data.
