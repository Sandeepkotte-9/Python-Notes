What is Unsupervised Learning?

Unsupervised Learning is a machine learning approach where the model learns patterns from unlabeled data. Unlike supervised learning, there are no predefined target labels.

Applications
Customer segmentation
Market basket analysis
Anomaly detection
Document clustering
Recommendation systems

Example: Given customer purchase data without any customer categories, the algorithm can automatically group similar customers together.

Clustering
Clustering is the process of grouping similar data points into clusters.

Goal

Similar points should belong to the same cluster.

Dissimilar points should belong to different clusters.

Example A shopping website groups customers based on buying behavior:

Cluster 1: Frequent buyers Cluster 2: Seasonal buyers Cluster 3: Premium customers

Hierarchical Clustering and it's types:
Hierarchical clustering builds a hierarchy of clusters.

Agglomerative Clustering (Bottom-Up) Starts with each data point as its own cluster. Gradually merges the closest clusters. Continues until all points belong to one cluster. Example
Initially: A B C D Merge closest clusters: (A,B) C D Then: (A,B,C) D Finally: (A,B,C,D)

Divisive Clustering (Top-Down) Starts with one large cluster. Repeatedly splits clusters into smaller groups. 
Example
Initially: (A,B,C,D) Split: (A,B) (C,D) Further split: A B C D

K-Means Clustering
As the name suggest, it is based on the mean, which is actually named as centroid and so K-Means is a centroid-based clustering algorithm.

How It Works:

1.Choose the number of clusters (K).

2.Randomly initialize K centroids.

3.Assign each point to the nearest centroid.

4.Recalculate centroids using the mean of assigned points.

5.Repeat until centroids stop changing significantly.

Example Suppose customer ages are: 18, 20, 22, 45, 48, 50 For K = 2 Cluster 1: 18, 20, 22 Cluster 2: 45, 48, 50 Centroids: 20 and 47.67

How to choose the Value of K:

Based on the classes we can decide the value of K.

Random K Method Try different K values and inspect the clusters. After inspect, we can decide, if we need to reduce or increase the number of K values.

Example:K = 2 K = 3 K = 4 Choosing the one that creates meaningful clusters.
Elbow Method
It is used to determine the optimal value of K.
How to do that?

Calculate:

WCSS (Within Cluster Sum of Squares): WCSS measures how compact the clusters are.

As K increases, WCSS decreases.

After a certain point, improvement becomes minimal, that point is called the Elbow Point.

Example 1. K WCSS 1 1000 2 600 3 300 4 250 5 220

The elbow appears around K = 3.

DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
DBSCAN forms clusters based on data density.
Unlike K-Means:
No need to specify K.
Can detect arbitrary-shaped clusters.
Handles noise effectively.
Important Terminology:

1.Epsilon (ε): Defines the radius around a point. Example: If ε = 2, all points within distance 2 are considered neighbors.

2.MinPts (Minimum Points): Minimum number of points required to form a dense region. Example: MinPts = 5, a point needs at least 5 neighboring points within ε distance.

3.Core Point: A point having at least MinPts neighbors within ε. Example: A crowded area in a city map.

4.Border Point: A point near a core point but does not satisfy MinPts itself. Example: A house located on the edge of a neighborhood.

5.Noise Point (Outlier): Neither a core point nor a border point. There are chances that the noise point can be turned into a core point or a border point. 
Example: An isolated house far away from all others.

When to Use DBSCAN?

Data is very dense.
Clusters have irregular shapes.
Dataset contains noise or outliers.
Example: Detecting crowded regions in GPS location data.

Cluster Evaluation
Used to measure clustering quality.

External Evaluation: Compares clustering results with actual labels.
Example: Comparing generated customer groups with known customer categories.
Internal Evaluation: Measures cluster quality using only the dataset.
Example: Checking whether points inside a cluster are close together.
Metrics:
Silhouette Score: Silhouette is a method of interpretation and validation of consistency within clusters of data.

Reliability / Stability: Checks whether clustering remains similar across multiple runs.

Example Running K-Means several times and comparing results.
