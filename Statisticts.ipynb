Statistics Notes
1. Statistics
Statistics is the science of collecting, organizing, analyzing, and interpreting data to make decisions or draw conclusions.

i) Descriptive Statistics — Summarizes and presents data using tables, graphs, and numerical measures. Example: The average score of a class is 82%.

ii) Inferential Statistics — Uses sample data to make conclusions or predictions about a larger population. Example: Polling 1,000 voters to estimate how all voters in a country will vote.

2. Population
The population is the entire group of individuals or items of interest in a study. Example: All students at a university.

```python
3. Sample
A sample is a subset of the population selected for study. Example: 200 students chosen from the university.
```

4. Variable
A variable is a characteristic that can take different values for different individuals. Example: Height, age, or eye color.

5. Data
Data are the observed values collected for variables.

Datum (singular): One observation. Example: One student's age is 20.
Data (plural): Multiple observations. Example: Ages of 200 students.

6. Experiment
An experiment is a study in which researchers impose a treatment or condition and observe its effects. Example: Testing whether a new fertilizer increases plant growth.

7. Parameter
A parameter is a numerical measure that describes a population. Example: The average age of all students at a university.

8. Statistic
A statistic is a numerical measure calculated from a sample. Example: The average age of the 200 sampled students.

Types of Data
1. Categorical Data — Consists of labels or categories rather than numerical measurements. Example: Gender, blood type, favorite color, car brand.

2. Numerical Data — Consists of numbers that represent counts or measurements.

a) Discrete Data — Countable values, usually whole numbers. Example: Number of children in a family, number of cars owned, number of defects in a product.
b) Continuous Data — Can take any value within a range and are usually obtained by measurement. Examples: Height (170.5 cm), weight (68.2 kg), temperature (22.7°C), time (3.45 seconds).

Levels of Measurement
i) Qualitative — A variable that categorizes or describes a population element.

ii) Quantitative — A variable that quantifies a population element (Nominal, Ordinal, Interval, Ratio).

1. Nominal Data — Consists of categories with no natural order or ranking. Example: Nationality, blood type.

2. Ordinal Data — Consists of categories that can be ranked, but the differences between ranks are not necessarily equal. Example: Class rank, education level.

3. Interval Data — Numerical data with equal intervals between values, but there is no true zero. Example: Temperature in °C or °F.

4. Ratio Data — Numerical data with equal intervals and a true zero point. Example: Height, age, income.



```python
Measures of Central Tendency
1. Mean — The mean (average) is the sum of all values divided by the number of values.

2. Median — The middle value when data are arranged in order. Example: 3, 5, 7, 9, 11 → Median = 7

3. Mode — The value that occurs most frequently. Example: 2, 3, 3, 4, 5 → Mode = 3
```

Measures of Dispersion
1. Range — The difference between the largest and smallest values.

2. Variance — Measures the average squared deviation of data values from the mean. Example: If scores are very spread out, the variance is large; if scores are clustered near the mean, the variance is small.

3. Standard Deviation — The square root of the variance. It measures the typical distance of data values from the mean. Example: If the mean test score is 80 and the standard deviation is 5, most scores tend to be within about 5 points of 80.

```python
Distribution Shape
1. Symmetrical Distribution — Data evenly distributed around the center. The left and right sides are approximately mirror images. Mean = Median = Mode.

2. Asymmetrical Distribution — Not balanced around its center. One side extends farther than the other. The distribution is skewed either left or right.

3. Right-Skewed Distribution — A long tail extending to the right (toward larger values).

4. Left-Skewed Distribution — A long tail extending to the left (toward smaller values).

5. Bell Curve — A perfectly symmetrical, bell-shaped distribution known as the normal distribution.
```

Distributions
1. Normal Distribution — A continuous, bell-shaped, symmetric distribution where most values cluster around the mean.

2. Bernoulli Distribution — Describes a single experiment with only two possible outcomes: success or failure. Example: Pass/fail.

3. Binomial Distribution — Describes the number of successes in a fixed number of independent Bernoulli trials. Example: Number of heads in 10 coin tosses.

4. Uniform Distribution — All outcomes are equally likely. Example: Chances of getting 1 when a die is rolled (1/6).

5. Poisson Distribution — Models the number of times an event occurs in a fixed interval of time, area, distance, or volume. Example: Number of phone calls received per hour.



Empirical Rule & Standardization
1) Empirical Rule (68–95–99.7 Rule)

For a normal distribution:

68% of observations lie within 1 standard deviation of the mean.
95% lie within 2 standard deviations of the mean.
99.7% lie within 3 standard deviations of the mean.
2) Standard Normal Distribution

The standard normal distribution is a special normal distribution with:

Mean (μ) = 0
Standard deviation (σ) = 1
A z-score indicates how many standard deviations a value is above or below the mean. Example: A z-score of +2 means a value is 2 standard deviations above the mean.

3) Standardization

Standardization is the process of converting a normal random variable X into a standard normal variable Z.
